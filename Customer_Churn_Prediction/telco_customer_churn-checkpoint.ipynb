# Project :- Customer Churn Prediction

## PART A – Step 1: Problem Framing & Theory Notes

### Q1. What is Customer Churn? Why is predicting churn valuable for a business like Jio or Airtel? (Mention CAC vs CLV.)

### 1. Customer Churn

Customer Churn means when a customer stops using a company's
products or services.

For example, when a Jio customer switches to Airtel,
it is called Customer Churn.

### Why is Customer Churn Prediction Important?

- It helps companies identify customers who may leave.
- Companies can offer discounts and better services to retain customers.
- It reduces customer loss and increases business profit.

### CAC vs CLV

- CAC (Customer Acquisition Cost): Money spent to get a new customer.
- CLV (Customer Lifetime Value): Total revenue expected from a customer
  during their relationship with the company.

Retaining existing customers is often cheaper than acquiring new customers.

### Q2. Define the four cells of a Confusion Matrix (TP, TN, FP, FN). In the churn context — what is the business cost of a False Negative? What about a False Positive?

### 2. Confusion Matrix

A Confusion Matrix is used to evaluate the performance
of a classification model.

- TP (True Positive): Actual customer churned and model predicted Churn.
- TN (True Negative): Actual customer stayed and model predicted No Churn.
- FP (False Positive): Actual customer stayed but model predicted Churn.
- FN (False Negative): Actual customer churned but model predicted No Churn.

### Business Cost

- False Negative: Company misses a customer who is actually leaving.
  This can cause loss of revenue.
- False Positive: Company contacts a customer who was not going to leave.
  This wastes time and retention resources.

## Q3. What is Class Imbalance? Does the Telco Churn dataset have it? How does SMOTE address it?

### 3. Class Imbalance

Class Imbalance occurs when one class has significantly
more samples than another class.

The Telco Customer Churn dataset is imbalanced because
No Churn customers are more than Churn customers.

### SMOTE

SMOTE stands for Synthetic Minority Over-sampling Technique.

- It creates synthetic samples for the minority class.
- It helps balance the training dataset.
- It can help the model learn minority class patterns better.

SMOTE should only be applied to training data, not test data.

## Q4. Briefly explain how each of the 4 algorithms you will use works — KNN, Naive Bayes, SVM, Decision Tree .

### 4. Classification Algorithms

#### 1. K-Nearest Neighbors (KNN)
KNN predicts the class of a new data point by checking
its nearest K neighbors. It assigns the majority class
among those neighbors.

#### 2. Naive Bayes
Naive Bayes is based on Bayes' Theorem.
It predicts the class using probability and assumes
that features are independent of each other.

#### 3. Support Vector Machine (SVM)
SVM finds the best decision boundary (hyperplane)
that separates different classes with the maximum margin.

#### 4. Decision Tree
Decision Tree makes predictions using a tree-like structure.
It splits data based on features and their conditions
until it reaches a final prediction.

## Q5. When would you choose Precision over Recall as your primary metric? Give a churn-specific answer.

### 5. Precision vs Recall

Precision is important when we want to reduce False Positives.

Recall is important when we want to identify as many actual
churn customers as possible.

In customer churn prediction, Recall is generally more important
because missing an actual churn customer can cause business loss.

However, if the company has limited resources and wants to contact
only highly likely churn customers, Precision can also be important.

StandardScaler kya karta hai?

StandardScaler numerical values ko ek common scale par convert karta hai.

Formula:
\[
Z = \frac{X-\mu}{\sigma}
\]

Where:
- X = Original value
- μ = Mean
- σ = Standard Deviation
- Z = Scaled value
Real-life example:

Suppose:
- Age = 25
- Salary = 50,000
Dono ke scales different hain. StandardScaler values ko comparable scale par laata hai.

Important exam concept:
- KNN = Scaling important.
- SVM = Scaling important.
- Decision Tree = Scaling generally required nahi.
- Gaussian Naive Bayes = Scaling generally required nahi.

# PART D – Step 4: Model Building (KNN & Naive Bayes)

## 4.1 – K-Nearest Neighbors (KNN)

### 4.1.1 – KNN Baseline Model (k=5)

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='median')

X_train_imputed = pd.DataFrame(
    imputer.fit_transform(X_train),
    columns=X_train.columns,
    index=X_train.index
)

X_test_imputed = pd.DataFrame(
    imputer.transform(X_test),
    columns=X_test.columns,
    index=X_test.index
)

print("Training NaN:", X_train_imputed.isnull().sum().sum())
print("Testing NaN:", X_test_imputed.isnull().sum().sum())
```

```python
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE

scaler = StandardScaler()

numeric_columns = [
    'tenure', 'MonthlyCharges', 'TotalCharges', 'num_services'
]

X_train_scaled = X_train_imputed.copy()
X_test_scaled = X_test_imputed.copy()

X_train_scaled[numeric_columns] = scaler.fit_transform(
    X_train_imputed[numeric_columns]
)

X_test_scaled[numeric_columns] = scaler.transform(
    X_test_imputed[numeric_columns]
)

# Apply SMOTE
smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_scaled, y_train
)

print("SMOTE completed!")
print("NaN in training data:", X_train_smote.isnull().sum().sum())
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)

knn.fit(X_train_smote, y_train_smote)

print("KNN Model Trained Successfully!")
```

```python
# Predict class labels
y_pred_knn = knn.predict(X_test_scaled)

# Predict probabilities
y_prob_knn = knn.predict_proba(X_test_scaled)[:, 1]

print("Predictions completed!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

print("KNN Model Performance")

print("Accuracy:", accuracy_score(y_test, y_pred_knn))
print("Precision:", precision_score(y_test, y_pred_knn))
print("Recall:", recall_score(y_test, y_pred_knn))
print("F1 Score:", f1_score(y_test, y_pred_knn))
print("AUC-ROC:", roc_auc_score(y_test, y_prob_knn))
```

```python
from sklearn.model_selection import cross_val_score
import matplotlib.pyplot as plt

k_values = [1, 3, 5, 7, 9, 11, 15]
f1_scores = []

for k in k_values:

    model = KNeighborsClassifier(n_neighbors=k)

    scores = cross_val_score(
        model,
        X_train_smote,
        y_train_smote,
        cv=3,
        scoring='f1'
    )

    f1_scores.append(scores.mean())

    print("K:", k, "F1 Score:", scores.mean())
```

```python
plt.figure(figsize=(8, 5))

plt.plot(k_values, f1_scores, marker='o')

plt.xlabel("K Values")
plt.ylabel("F1 Score")
plt.title("KNN - Tuning K Values")

plt.show()
```

```python
best_k = k_values[f1_scores.index(max(f1_scores))]

print("Best K Value:", best_k)
print("Best F1 Score:", max(f1_scores))
```

### 4.1.3 – Confusion Matrix

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

best_knn = KNeighborsClassifier(n_neighbors=best_k)

best_knn.fit(X_train_smote, y_train_smote)

y_pred_best_knn = best_knn.predict(X_test_scaled)

cm = confusion_matrix(y_test, y_pred_best_knn)

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['No Churn', 'Churn']
).plot(cmap='Blues')

plt.title("KNN Confusion Matrix")
plt.show()
```

### 4.1.4 – ROC Curve

```python
from sklearn.metrics import roc_curve, roc_auc_score

y_prob_best_knn = best_knn.predict_proba(X_test_scaled)[:, 1]

fpr, tpr, thresholds = roc_curve(y_test, y_prob_best_knn)

plt.figure(figsize=(7, 5))

plt.plot(fpr, tpr, label='KNN')

plt.plot([0, 1], [0, 1], linestyle='--')

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("KNN ROC Curve")

plt.legend()
plt.show()

print("KNN AUC:", roc_auc_score(y_test, y_prob_best_knn))
```

### 5.1.4 – Hyperparameter Tuning (C Value)

```python
from sklearn.model_selection import cross_val_score
from sklearn.svm import SVC

C_values = [0.1, 1, 10, 100]
svm_f1_scores = []

for c in C_values:

    model = SVC(
        kernel='rbf',
        C=c,
        gamma='scale'
    )

    scores = cross_val_score(
        model,
        X_train_smote,
        y_train_smote,
        cv=3,
        scoring='f1'
    )

    svm_f1_scores.append(scores.mean())

    print("C:", c, "F1 Score:", scores.mean())
```

```python
best_C = C_values[svm_f1_scores.index(max(svm_f1_scores))]

print("Best C Value:", best_C)
print("Best F1 Score:", max(svm_f1_scores))
```

### 5.1.5 – Train Best SVM Model

```python
best_svm = SVC(
    kernel='rbf',
    C=best_C,
    gamma='scale',
    probability=True,
    random_state=42
)

best_svm.fit(X_train_smote, y_train_smote)

y_pred_best_svm = best_svm.predict(X_test_scaled)
y_prob_best_svm = best_svm.predict_proba(X_test_scaled)[:, 1]

print("Best SVM Performance")

print("Accuracy:", accuracy_score(y_test, y_pred_best_svm))
print("Precision:", precision_score(y_test, y_pred_best_svm))
print("Recall:", recall_score(y_test, y_pred_best_svm))
print("F1 Score:", f1_score(y_test, y_pred_best_svm))
print("AUC-ROC:", roc_auc_score(y_test, y_prob_best_svm))
```

# PART F – Step 6: Model Evaluation & Comparison

### 6.1 – Consolidated Comparison Table

```python
import time
import pandas as pd

from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# Create all 4 best models
models = {
    "KNN": KNeighborsClassifier(n_neighbors=best_k),

    "Naive Bayes": GaussianNB(),

    "SVM": SVC(
        kernel='rbf',
        C=best_C,
        gamma='scale',
        probability=True,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=best_depth,
        class_weight='balanced',
        random_state=42
    )
}

results = []
trained_models = {}

# Train and evaluate each model
for name, model in models.items():

    # Calculate training time
    start_time = time.time()

    model.fit(X_train_smote, y_train_smote)

    end_time = time.time()

    training_time = end_time - start_time

    # Predictions
    y_pred = model.predict(X_test_scaled)
    y_prob = model.predict_proba(X_test_scaled)[:, 1]

    # Evaluation metrics
    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred),
        'Recall': recall_score(y_test, y_pred),
        'F1-Score': f1_score(y_test, y_pred),
        'AUC-ROC': roc_auc_score(y_test, y_prob),
        'Training Time (seconds)': training_time
    })

    trained_models[name] = model

# Create DataFrame
comparison_df = pd.DataFrame(results)

# Sort by Recall in descending order
comparison_df = comparison_df.sort_values(
    by='Recall',
    ascending=False
)

print(comparison_df.round(4))
```

### 6.2 – Plot All 4 ROC Curves in One Figure

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, roc_auc_score

plt.figure(figsize=(9, 6))

for name, model in trained_models.items():

    y_prob = model.predict_proba(X_test_scaled)[:, 1]

    fpr, tpr, thresholds = roc_curve(y_test, y_prob)

    auc = roc_auc_score(y_test, y_prob)

    plt.plot(
        fpr,
        tpr,
        label=f'{name} (AUC = {auc:.3f})'
    )

# Random prediction reference line
plt.plot(
    [0, 1],
    [0, 1],
    linestyle='--',
    label='Random Classifier'
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison - All 4 Models")

plt.legend()
plt.grid()
plt.show()
```
